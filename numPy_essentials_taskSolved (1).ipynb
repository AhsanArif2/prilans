### 1. Normal distribution

$$f(x) = \frac{1}{\sigma\sqrt{2\pi}} \exp\left(-\frac{(x - \mu)^2}{2\sigma^2}\right)$$

Compute the density at every value of `x`, with mean $\mu = 0$ and standard deviation $\sigma = 1$. Use `np.pi`.

Expected (rounded): `[0.054, 0.242, 0.399, 0.242, 0.054]`.

```python
import numpy as np

x = np.array([-2.0, -1.0, 0.0, 1.0, 2.0])
mu, sigma = 0.0, 1.0
density = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-((x - mu) ** 2) / (2 * sigma ** 2))
print(np.round(density, 3))
```

### 2. Standardisation (z-score)

$$z_{ij} = \frac{x_{ij} - \mu_j}{\sigma_j}$$

Each row of `X` is one person and each column is one measurement. $\mu_j$ and $\sigma_j$ are the mean and standard deviation of column $j$.
After standardising, every column should have mean 0 and standard deviation 1.

Expected: `[0., 0.]` and `[1., 1.]`.

```python
X = np.array([[150, 50],
              [160, 60],
              [170, 65],
              [180, 85]], dtype=float)
Z = (X - X.mean(axis=0)) / X.std(axis=0)
print(np.round(Z.mean(axis=0), 6))
print(Z.std(axis=0))
```

### 3. Pearson correlation

$$r = \frac{\sum_{i=1}^{n}(x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum_{i=1}^{n}(x_i - \bar{x})^2}\,\sqrt{\sum_{i=1}^{n}(y_i - \bar{y})^2}}$$

$r$ is close to 1 when $y$ rises steadily with $x$. Check your answer against `np.corrcoef(x, y)[0, 1]`.

Expected: about `0.9815`.

```python
hours = np.array([1, 2, 3, 4, 5], dtype=float)
exam_marks = np.array([52, 55, 61, 70, 72], dtype=float)
r = np.sum((hours - hours.mean()) * (exam_marks - exam_marks.mean())) / np.sqrt(
    np.sum((hours - hours.mean()) ** 2) *
    np.sum((exam_marks - exam_marks.mean()) ** 2)
)
print(r)
print(np.corrcoef(hours, exam_marks)[0, 1])
```

### 5. Sigmoid and binary cross-entropy

$$p_i = \frac{1}{1 + e^{-z_i}}, \qquad L = -\frac{1}{n}\sum_{i=1}^{n}\Big[y_i \log p_i + (1 - y_i)\log(1 - p_i)\Big]$$

$p_i$ is the predicted probability that example $i$ belongs to class 1, and $y_i$ is its true label, 0 or 1.

Expected: `p` about `[0.119, 0.5, 0.953]` and `L` about `0.2895`.

```python
z = np.array([-2.0, 0.0, 3.0])
y = np.array([0, 1, 1])
p = 1 / (1 + np.exp(-z))
eps = 1e-15
p_safe = np.clip(p, eps, 1 - eps)
bce = -np.mean(y * np.log(p_safe) + (1 - y) * np.log(1 - p_safe))
print(np.round(p, 3))
print(bce)
```

### 6. Softmax

$$s_{ij} = \frac{e^{z_{ij} - m_i}}{\sum_{k} e^{z_{ik} - m_i}}, \qquad m_i = \max_k z_{ik}$$

Softmax turns each row of scores into probabilities that add up to 1. Subtracting the row maximum $m_i$ does not change the result,
but it stops `np.exp` from overflowing on large scores like the second row.

Expected rows (rounded): `[0.09, 0.245, 0.665]` and `[0.333, 0.333, 0.333]`.

```python
scores = np.array([[1.0, 2.0, 3.0],
                   [1000.0, 1000.0, 1000.0]])
scores_shifted = scores - np.max(scores, axis=1, keepdims=True)
exp_scores = np.exp(scores_shifted)
softmax = exp_scores / np.sum(exp_scores, axis=1, keepdims=True)
print(np.round(softmax, 3))
print(softmax.sum(axis=1))
```

**Solution run:**

```python
scores = np.array([[1.0, 2.0, 3.0],
                   [1000.0, 1000.0, 1000.0]])
softmax = scores_shifted = scores - np.max(scores, axis=1, keepdims=True)
exp_scores = np.exp(scores_shifted)
softmax = exp_scores / np.sum(exp_scores, axis=1, keepdims=True)
print(np.round(softmax, 3))
print(softmax.sum(axis=1))
```

### 8. Cross-entropy loss

$$L = -\frac{1}{n}\sum_{i=1}^{n} \log P_{i, y_i}$$

Row $i$ of `P` holds the predicted probabilities for example $i$, and $y_i$ is its true class.
Select each row's probability at its true class with one index, then apply the formula.

Expected: about `0.4987`.

```python
P = np.array([[0.7, 0.2, 0.1],
              [0.1, 0.8, 0.1],
              [0.3, 0.3, 0.4]])
true_classes = np.array([0, 1, 2])
cross_entropy = -np.mean(np.log(P[np.arange(len(true_classes)), true_classes]))
print(cross_entropy)
```
